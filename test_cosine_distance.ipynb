```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_distances
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import squareform

import plotly.graph_objects as go

import utility_functions as uf
```

```python
path = "data/"
```

```python
df_country_subfield = (
    pd.read_csv(path+"df_country_subfield_norm_world_log.csv", index_col=0)
    .dropna(how="all", axis=1)
    .fillna(0)
)
```

```python
len(df_country_subfield.columns)
```

```python
df_country_subfield
```

```python
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(df_country_subfield.values)
```

```python
df_country_subfield.values.mean()
```

```python
X_scaled.mean()
```

```python
D = cosine_distances(df_country_subfield.values)
D_scaled = cosine_distances(X_scaled)
df_country_dist_cosine = pd.DataFrame(D,
                                     index=df_country_subfield.index,
                                     columns=df_country_subfield.index)
df_country_dist_cosine_scaled = pd.DataFrame(D_scaled,
                                     index=df_country_subfield.index,
                                     columns=df_country_subfield.index)
df_country_dist_cosine.to_csv(path+"df_country_dist_cosine_world_log.csv")
# df_country_dist_cosine.to_csv(path+"df_country_dist_cosine_scaled_world.csv")
```

```python
# x_labels = [uf.id2name_country[country] for country in (map(str, df_country_dist_cosine.columns))]
# y_labels = [uf.id2name_country[country] for country in (map(str, df_country_dist_cosine.columns))]

sub_list = list(map(str, uf.top_n_countries_by_articles(20)))

# country = "AX"
# top_n = 20
# sub_list = df_country_dist_w1.sort_values(country, ascending=True)[[country]].head(top_n).index.to_list()

x_labels = sub_list
y_labels = sub_list

fig = go.Figure(
    data=go.Heatmap(
        # z=df_country_dist_cosine.values,
        z=df_country_dist_cosine.loc[sub_list, sub_list].values,
        x=x_labels,
        y=y_labels,
        colorscale="Viridis",
        zmin=0,
        zmax=1,
    )
)

# Layout
fig.update_layout(
    title="Distances for top-20 countries",
    width=1200,
    # height=20*len(y_labels),  # taller if more rows
    height=40*len(y_labels),  # taller if more rows
    margin=dict(l=100, r=50, t=50, b=100),
    xaxis=dict(tickangle=45, automargin=True),
    yaxis=dict(autorange="reversed", automargin=True)  # keep top-to-bottom ordering
)

fig.show()
```
