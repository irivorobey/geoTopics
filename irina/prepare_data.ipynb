```python
import pandas as pd
import numpy as np

import utility_functions as uf
```

```python
mode_list = ["individual", "collaborative"]
```

```python
df_cs_yearly_list = []

start_year = 1970
end_year = 2023
for mode in mode_list:
    for year_loop in range(start_year, end_year+1):
        df_cs_loop = (
            pd.read_csv(uf.PATH+f"df_country_subfield/{mode}_{year_loop}.csv", index_col=0)
            .assign(year=year_loop, mode=mode)
            .reset_index(names="country")
        )
        df_cs_yearly_list.append(df_cs_loop)
df_cs_yearly_full = pd.concat(df_cs_yearly_list).set_index(["country", "year", "mode"])
df_prob_yearly_full = df_cs_yearly_full.div(df_cs_yearly_full.sum(axis=1), axis=0)
```

```python
df_cs_yearly_full.to_csv(uf.PATH+f"df_cs_yearly.csv")
```
