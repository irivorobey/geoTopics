```python
def get_base(df_dist, df_stats, c_list, grouped_list):
    non_grouped = [c for c in df_dist.index if (c not in grouped_list) and (c not in c_list)]
    countries_sorted = (
        df_stats
        .loc[non_grouped]
        .sort_values(by=["total_articles", "norm_entropy"], ascending=False)
        .head(10)
        .index
        .to_list()
    )
    display(
        df_dist
        .loc[countries_sorted, countries_sorted]
        .max(axis=1)
        .to_frame("dist_max")
        .assign(
            c_max = df_dist.loc[countries_sorted, countries_sorted].apply(lambda row: uf.top_n_columns(row, 1), axis=1),
        )
        .sort_values("dist_max", ascending=False)
    )

    return
```
