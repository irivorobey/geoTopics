```python
def get_base(df_dist, df_stats, c_list, grouped_list):
    non_grouped = [c for c in df_dist.index if (c not in grouped_list) and (c not in c_list)]
    countries_sorted = (
        df_stats
        .loc[non_grouped]
        .sort_values(by=["total_articles", "norm_entropy"], ascending=False)
        .head(10)
        .index
        .to_list()
    )
    df_tmp = (
        df_dist
        .loc[countries_sorted, countries_sorted]
        .max(axis=1)
        .to_frame("dist_max")
        .assign(
            c_max = df_dist.loc[countries_sorted, countries_sorted].apply(lambda row: uf.top_n_columns(row, 1), axis=1),
        )
        .sort_values("dist_max", ascending=False)
    )
    display(countries_sorted)
    print([df_tmp.c_max.iloc[0]])
    return [df_tmp.c_max.iloc[0]]
```

```python
non_grouped = df_dist.index.to_list()
grouped = []
c_list = []
group_dict = {}

n = 10

while non_grouped:
    c_list += get_base(df_dist, df_stats, c_list, grouped)
    non_grouped = [c for c in df_dist.index if (c not in grouped) and (c not in c_list)]
    for c in c_list:
        tmp_list = df_dist.loc[[c], non_grouped].apply(lambda row: uf.bottom_n_columns(row, n), axis=1).iloc[0]
        if c in group_dict.keys():
            group_dict[c] += tmp_list
        else:
            group_dict[c] = tmp_list + [c]
        grouped += tmp_list
    non_grouped = [c for c in df_dist.index if (c not in grouped) and (c not in c_list)]
```
