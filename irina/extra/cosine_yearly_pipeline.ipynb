# Pipeline

```python
path = uf.PATH + "df_country_subfield/"

df_country_stats_list = []
df_subfield_stats_collab_list = []
df_subfield_stats_individual_list = []
df_interest_metric_stats_collab_list = []
df_interest_metric_stats_individual_list = []
df_distance_stats_collab_list = []
df_distance_stats_individual_list = []

for year in tqdm(np.arange(1970, 2024)):
    df_country_subfield_collab = pd.read_csv(path+""+str(year)+".csv", index_col=0)
    df_country_subfield_individual = pd.read_csv(path+"individual_"+str(year)+".csv", index_col=0)

    df_country_stats_collab = uf.get_country_stats(df_country_subfield_collab)
    df_country_stats_individual = uf.get_country_stats(df_country_subfield_individual)

    col_list_coupled = [[col+"_collab", col+"_individual"] for col in df_country_stats_collab.columns.to_list()]
    col_list = [x for sublist in col_list_coupled for x in sublist]

    df_country_stats_compare = (
        df_country_stats_collab
        .merge(df_country_stats_individual, left_index=True, right_index=True, how="outer", suffixes=("_collab", "_individual"))
        [col_list]
        .assign(year=year)
        .reset_index()
    )

    df_interest_metric_collab = uf.get_interest_metric(df_country_subfield_collab)
    df_interest_metric_individual = uf.get_interest_metric(df_country_subfield_individual)

    df_subfield_stats_collab = uf.get_subfield_stats(df_country_subfield_collab, df_interest_metric_collab, year)
    df_subfield_stats_individual = uf.get_subfield_stats(df_country_subfield_individual, df_interest_metric_individual, year)

    df_interest_metric_stats_collab = uf.get_interest_metric_stats(df_interest_metric_collab, df_subfield_stats_collab, year)
    df_interest_metric_stats_individual = uf.get_interest_metric_stats(df_interest_metric_individual, df_subfield_stats_individual, year)

    df_distance_collab = uf.get_cosine_distances(df_interest_metric_collab.fillna(0))
    df_distance_individual = uf.get_cosine_distances(df_interest_metric_individual.fillna(0))

    df_distance_stats_collab = uf.get_distance_stats(df_distance_collab, year=year)
    df_distance_stats_individual = uf.get_distance_stats(df_distance_individual, year=year)

    df_country_stats_list.append(df_country_stats_compare)
    df_subfield_stats_collab_list.append(df_subfield_stats_collab)
    df_subfield_stats_individual_list.append(df_subfield_stats_individual)
    df_interest_metric_stats_collab_list.append(df_interest_metric_stats_collab)
    df_interest_metric_stats_individual_list.append(df_interest_metric_stats_individual)
    df_distance_stats_collab_list.append(df_distance_stats_collab)
    df_distance_stats_individual_list.append(df_distance_stats_individual)

    df_interest_metric_collab.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric/collab_{year}.csv")
    df_interest_metric_individual.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric/individual_{year}.csv")
    df_distance_collab.to_csv(uf.PATH+f"cosine_pipeline/df_distance/collab_{year}.csv")
    df_distance_individual.to_csv(uf.PATH+f"cosine_pipeline/df_distance/individual_{year}.csv")

df_country_stats_compare = pd.concat(df_country_stats_list)
df_subfield_stats_collab = pd.concat(df_subfield_stats_collab_list)
df_subfield_stats_individual = pd.concat(df_subfield_stats_individual_list)
df_interest_metric_stats_collab = pd.concat(df_interest_metric_stats_collab_list)
df_interest_metric_stats_individual = pd.concat(df_interest_metric_stats_individual_list)
df_distance_stats_collab = pd.concat(df_distance_stats_collab_list)
df_distance_stats_individual = pd.concat(df_distance_stats_individual_list)

df_country_stats_compare.to_csv(uf.PATH+f"cosine_pipeline/df_country_stats_compare.csv", index=False)
df_subfield_stats_collab.to_csv(uf.PATH+f"cosine_pipeline/df_subfield_stats_collab.csv", index=False)
df_subfield_stats_individual.to_csv(uf.PATH+f"cosine_pipeline/df_subfield_stats_individual.csv", index=False)
df_interest_metric_stats_collab.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric_stats_collab.csv", index=False)
df_interest_metric_stats_individual.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric_stats_individual.csv", index=False)
df_distance_stats_collab.to_csv(uf.PATH+f"cosine_pipeline/df_distance_stats_collab.csv", index=False)
df_distance_stats_individual.to_csv(uf.PATH+f"cosine_pipeline/df_distance_stats_individual.csv", index=False)
```
