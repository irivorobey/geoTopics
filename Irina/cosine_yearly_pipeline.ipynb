# Pipeline

```python
np.arange(2000, 2024)
```

```python
path = uf.PATH + "df_country_subfield_Floriana/"

df_country_stats_list = []
df_subfield_stats_list = []
df_subfield_stats_unique_list = []
df_interest_metric_stats_list = []
df_interest_metric_stats_unique_list = []
df_distance_stats_list = []
df_distance_stats_unique_list = []

for year in tqdm(np.arange(2000, 2024)):
    df_country_subfield = pd.read_csv(path+"country_subfield_matrix_"+str(year)+".csv", index_col=0)
    df_country_subfield_unique = pd.read_csv(path+"country_subfield_matrix_SingleCountry_"+str(year)+".csv", index_col=0)

    df_country_stats = uf.get_country_stats(df_country_subfield)
    df_country_stats_unique = uf.get_country_stats(df_country_subfield_unique)

    col_list_coupled = [[col, col+"_unique"] for col in df_country_stats.columns.to_list()]
    col_list = [x for sublist in col_list_coupled for x in sublist]

    df_country_stats_compare = (
        df_country_stats
        .merge(df_country_stats_unique, left_index=True, right_index=True, how="outer", suffixes=("", "_unique"))
        [col_list]
        .assign(year=year)
        .reset_index()
    )

    df_interest_metric = uf.get_interest_metric(df_country_subfield)
    df_interest_metric_unique = uf.get_interest_metric(df_country_subfield_unique)

    df_subfield_stats = uf.get_subfield_stats(df_country_subfield, df_interest_metric, year)
    df_subfield_stats_unique = uf.get_subfield_stats(df_country_subfield_unique, df_interest_metric_unique, year)

    df_interest_metric_stats = uf.get_interest_metric_stats(df_interest_metric, df_subfield_stats, year)
    df_interest_metric_stats_unique = uf.get_interest_metric_stats(df_interest_metric_unique, df_subfield_stats_unique, year)

    df_distance = uf.get_cosine_distances(df_interest_metric.fillna(0))
    df_distance_unique = uf.get_cosine_distances(df_interest_metric_unique.fillna(0))

    df_distance_stats = uf.get_distance_stats(df_distance, year=year)
    df_distance_stats_unique = uf.get_distance_stats(df_distance_unique, year=year)

    df_country_stats_list.append(df_country_stats_compare)
    df_subfield_stats_list.append(df_subfield_stats)
    df_subfield_stats_unique_list.append(df_subfield_stats_unique)
    df_interest_metric_stats_list.append(df_interest_metric_stats)
    df_interest_metric_stats_unique_list.append(df_interest_metric_stats_unique)
    df_distance_stats_list.append(df_distance_stats)
    df_distance_stats_unique_list.append(df_distance_stats_unique)

    df_interest_metric.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric/{year}.csv")
    df_interest_metric_unique.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric/unique_{year}.csv")
    df_distance.to_csv(uf.PATH+f"cosine_pipeline/df_distance/{year}.csv")
    df_distance_unique.to_csv(uf.PATH+f"cosine_pipeline/df_distance/unique_{year}.csv")

df_country_stats_compare = pd.concat(df_country_stats_list)
df_subfield_stats = pd.concat(df_subfield_stats_list)
df_subfield_stats_unique = pd.concat(df_subfield_stats_unique_list)
df_interest_metric_stats = pd.concat(df_interest_metric_stats_list)
df_interest_metric_stats_unique = pd.concat(df_interest_metric_stats_unique_list)
df_distance_stats = pd.concat(df_distance_stats_list)
df_distance_stats_unique = pd.concat(df_distance_stats_unique_list)

df_country_stats_compare.to_csv(uf.PATH+f"cosine_pipeline/df_country_stats_compare.csv", index=False)
df_subfield_stats.to_csv(uf.PATH+f"cosine_pipeline/df_subfield_stats.csv", index=False)
df_subfield_stats_unique.to_csv(uf.PATH+f"cosine_pipeline/df_subfield_stats_unique.csv", index=False)
df_interest_metric_stats.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric_stats.csv", index=False)
df_interest_metric_stats_unique.to_csv(uf.PATH+f"cosine_pipeline/df_interest_metric_stats_unique.csv", index=False)
df_distance_stats.to_csv(uf.PATH+f"cosine_pipeline/df_distance_stats.csv", index=False)
df_distance_stats_unique.to_csv(uf.PATH+f"cosine_pipeline/df_distance_stats_unique.csv", index=False)
```
