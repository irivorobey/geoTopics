```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_distances
# from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
# from scipy.spatial.distance import squareform

import plotly.graph_objects as go

from Irina import utility_functions as uf
```

```python
path = "../data/"
```

```python
df_country_subfield = (
    pd.read_csv(path+"df_country_subfield_norm_world_log.csv", index_col=0)
    .dropna(how="all", axis=1)
    .fillna(0)
)
```

```python
df_country_subfield
```

```python
D = cosine_distances(df_country_subfield.values)

df_country_dist_cosine = pd.DataFrame(D,
                                     index=df_country_subfield.index,
                                     columns=df_country_subfield.index)

# df_country_dist_cosine.to_csv(path+"df_country_dist_cosine_world_log.csv")
```

```python
subset = uf.top_n_countries_by_articles(30)

fig = uf.plotly_heatmap(
    # df_country_dist_cosine,
    # y_labels=df_country_dist_cosine.index.astype(str).to_list(),
    # x_labels=df_country_dist_cosine.columns.astype(str).to_list(),
    df_country_dist_cosine.loc[subset, subset],
    y_labels=subset,
    x_labels=subset,
    y_type="country",
    x_type="country",
    y_name="X-Country",
    x_name="Y-Country",
    title="Cosine distance",
    line_height=30,
    z_min=0, z_max=2
)
```

```python
# country_list = ["US", "CN", "GB"]  # list of countries
country_list = ["US", "FR", "GB"]  # list of countries
x = df_country_subfield.columns.to_list()

fig = go.Figure()

for country in country_list:
    y = df_country_subfield.loc[country].values
    y_safe = [v if not np.isnan(v) else None for v in y]  # handle NaNs

    fig.add_trace(go.Scatter(
        x=x,
        y=y_safe,
        mode="lines",
        name=f"{uf.id2name_country[country]}"
    ))

fig.update_layout(
    title="Interest distributions for selected countries",
    xaxis=dict(title="Sub-field"),
    yaxis=dict(title="Sub-field share"),
    legend=dict(x=0.01, y=0.99),
    margin=dict(l=60, r=60, t=40, b=40)
)

fig.show()
```
