```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

import utility_functions as uf
```

```python

path = "data/"

df_country_subfield = pd.read_csv(path+"df_country_subfield.csv", index_col="country")
```

```python
df_country_subfield
```

```python
df_country_subfield_norm = (
    df_country_subfield
    .div(df_country_subfield.sum(axis=1), axis=0)
)
```

```python
df_world_subfield = (
    df_country_subfield.sum(axis=0)
    .to_frame()
    .T
)

df_world_subfield_norm = (
    df_world_subfield
    .div(df_world_subfield.sum(axis=1), axis=0)
)
df_world_subfield_norm
```

```python
df_country_subfield_norm_world = (
    df_country_subfield_norm
    .div(df_world_subfield_norm.loc[0])
)
df_country_subfield_norm_world_norm = (
    df_country_subfield_norm_world
    .div(df_country_subfield_norm_world.sum(axis=1), axis=0)
)
df_country_subfield_norm_world
```

```python
country = "FR"
```

```python
# df_country_subfield_norm_world_norm.to_csv(path+"df_country_subfield_norm_world.csv")
```

```python
df_plot = pd.concat([
    df_country_subfield_norm.loc[country],
    df_world_subfield_norm.loc[0],
    # df_country_subfield_norm_world.loc[country],
    df_country_subfield_norm_world_norm.loc[country]
], axis=1)
df_plot.columns = ["share_norm", "share_norm_world", "representation_norm"]
# df_plot
```

```python
x = df_country_subfield_norm.columns.to_list()

y1 = df_country_subfield_norm.loc[country]          # left axis
y2 = df_world_subfield_norm.loc[0]      # left axis
y3 = df_country_subfield_norm_world.loc[country]    # right axis (different scale)

fig = go.Figure()

# Left y-axis lines
fig.add_trace(
    go.Scatter(
        x=x,
        y=y1,
        mode="lines",
        name="Sub-field share across the country (left)",

    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=y2,
        mode="lines",
        name="Sub-field share across the world (left)",

    )
)

# Right y-axis line
fig.add_trace(
    go.Bar(
        x=x,
        y=y3,
        name="Subfield representation (right)",
        yaxis="y2",
        marker=dict(
            color="grey",
            opacity=0.4
        )
    )
)

fig.add_trace(
    go.Scatter(
        x=x,                         # same x-range as other traces
        y=[1] * len(x),              # constant line at y = 1
        mode="lines",
        name="Representation = 1 (right)",
        line=dict(color="grey", dash="dash"),
        showlegend=True,
        yaxis="y2",
    )
)

# Layout with secondary y-axis
fig.update_layout(
    title=uf.get_country_info(country).iloc[0, 0],
    xaxis=dict(title="Sub-field"),
    yaxis=dict(title="Sub-field share", overlaying="y2",),
    yaxis2=dict(
        title="Representation (1 = as in the world)",

        side="right"
    ),
    legend=dict(x=0.01, y=0.99),
    margin=dict(l=60, r=60, t=40, b=40)
)

fig.show()
```

```python

countries = ["FR", "ES", "SI", "PT", "AE", "IT", "BE", "HR", "LV"]
# Create figure
fig = go.Figure()

for country in countries:
    fig.add_trace(
        go.Scatter(
            x=df_country_subfield_norm_world_norm.index,
            y=df_country_subfield_norm_world_norm.loc[country].values,
            name=country,
            orientation='h',
            mode='lines',
        )
    )

fig.add_trace(
    go.Scatter(
        x=df_country_subfield_norm_world_norm.index,
        y=df_country_subfield_norm_world_norm.loc[countries].mean().values,
        name="Mean",
        orientation='h',
        mode='lines',
    )
)

# Layout
fig.update_layout(
    barmode='group',       # 'stack' if you prefer stacked bars
    title=f"Distribution",
    xaxis_title="Value",
    yaxis_title=None,
    margin=dict(l=150),    # leave space for long y labels
)

# Optional: largest bar at top
fig.update_yaxes(autorange="reversed")

fig.show()
```
